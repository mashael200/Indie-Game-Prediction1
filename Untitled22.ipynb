```python
import pandas as pd

df = pd.read_csv(r"C:\Users\96657\Downloads\SteamGames.csv")

df.drop_duplicates(inplace=True)

df.dropna(thresh=10, inplace=True)

df['release_date'] = pd.to_datetime(df['release_date'], errors='coerce')

def convert_owners(x):
    try:
        low, high = x.split('-')
        return (int(low) + int(high)) // 2
    except:
        return None

df['owners_clean'] = df['owners'].apply(convert_owners)

df['steamspy_tags'] = df['steamspy_tags'].astype(str)
df['genres'] = df['genres'].astype(str)

df_clean = df[['name','price','genres','release_date','positive_ratings',
               'negative_ratings','steamspy_tags','developer','publisher',
               'average_playtime','median_playtime','owners_clean']]

df_clean.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, ConfusionMatrixDisplay, classification_report
)
import matplotlib.pyplot as plt
import pandas as pd

# -----------------------------
# دالة تصنيف النجاح
# -----------------------------
def success_category(owners):
    if owners < 250_000:
        return "Low Success"
    elif owners < 1_000_000:
        return "Medium Success"
    else:
        return "High Success"

# -----------------------------
# تجهيز البيانات للتصنيف
# -----------------------------
classification_data = df_clean.copy()
classification_data["success_category"] = classification_data["owners_clean"].apply(success_category)

FEATURES = [
    "price",
    "positive_ratings",
    "negative_ratings",
    "average_playtime",
    "median_playtime",
    "owners_clean"
]

X_class = classification_data[FEATURES]
y_class = classification_data["success_category"]

# -----------------------------
# تقسيم البيانات
# -----------------------------
X_train_class, X_test_class, y_train_class, y_test_class = train_test_split(
    X_class,
    y_class,
    test_size=0.20,
    random_state=42,
    stratify=y_class
)

# -----------------------------
# Decision Tree
# -----------------------------
dt_class = DecisionTreeClassifier(
    random_state=42,
    max_depth=12
)
dt_class.fit(X_train_class, y_train_class)
y_pred_dt_class = dt_class.predict(X_test_class)

# -----------------------------
# Random Forest
# -----------------------------
rf_class = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)
rf_class.fit(X_train_class, y_train_class)
y_pred_rf_class = rf_class.predict(X_test_class)

# -----------------------------
# دوال التقييم
# -----------------------------
def classification_metrics(name, y_true, y_pred):
    return {
        "Model": name,
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred, average="weighted", zero_division=0),
        "Recall": recall_score(y_true, y_pred, average="weighted", zero_division=0),
        "F1-score": f1_score(y_true, y_pred, average="weighted", zero_division=0)
    }

results_class = pd.DataFrame([
    classification_metrics("Decision Tree", y_test_class, y_pred_dt_class),
    classification_metrics("Random Forest", y_test_class, y_pred_rf_class)
])

display(results_class)

# -----------------------------
# اختيار أفضل نموذج
# -----------------------------
best_classifier_name = (
    "Random Forest"
    if results_class.loc[results_class["Model"] == "Random Forest", "F1-score"].iloc[0]
       >= results_class.loc[results_class["Model"] == "Decision Tree", "F1-score"].iloc[0]
    else "Decision Tree"
)

best_classifier = rf_class if best_classifier_name == "Random Forest" else dt_class
best_predictions = y_pred_rf_class if best_classifier_name == "Random Forest" else y_pred_dt_class

print("Best Classifier:", best_classifier_name)
print("\nClassification Report:")
print(classification_report(y_test_class, best_predictions, zero_division=0))

# -----------------------------
# Confusion Matrix
# -----------------------------
labels = ["Low Success", "Medium Success", "High Success"]

cm = confusion_matrix(
    y_test_class,
    best_predictions,
    labels=labels
)

fig, ax = plt.subplots(figsize=(8, 6))
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=labels
)
disp.plot(ax=ax, cmap="Blues", values_format="d", colorbar=False)

ax.set_title(f"Confusion Matrix - {best_classifier_name}")
ax.set_xlabel("Predicted Label")
ax.set_ylabel("True Label")
plt.tight_layout()

plt.savefig("confusion_matrix.png", dpi=300, bbox_inches="tight")
plt.show()

print("Confusion Matrix saved as: confusion_matrix.png")
```

```python
classifier_importance = pd.Series(
    best_classifier.feature_importances_,
    index=FEATURES
).sort_values(ascending=True)

plt.figure(figsize=(10, 7))
plt.barh(classifier_importance.index, classifier_importance.values)
plt.title(f"{best_classifier_name} Feature Importance")
plt.xlabel("Importance")
plt.ylabel("Features")
plt.tight_layout()
plt.savefig("classifier_feature_importance.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

dt = DecisionTreeRegressor(random_state=42)
dt.fit(X_train, y_train)

y_pred_dt = dt.predict(X_test)

print("🌳 Decision Tree Results")
print("R²:", r2_score(y_test, y_pred_dt))
print("MAE:", mean_absolute_error(y_test, y_pred_dt))
print("MSE:", mean_squared_error(y_test, y_pred_dt))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred_dt)))

dt_importance = pd.Series(dt.feature_importances_, index=X.columns)
print("\n📌 Decision Tree Feature Importance:")
print(dt_importance.sort_values(ascending=False))
```
